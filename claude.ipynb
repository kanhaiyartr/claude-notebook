```python
from dotenv import load_dotenv
from pathlib import Path
from anthropic import Anthropic
import pprint

load_dotenv()
client = Anthropic()
model = "claude-sonnet-5-5"

response= client.messages.create(
    model=model,
    max_tokens=16000,
    system= SYSTEM_PROMPT,
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "type": "text",
                    "text": "Generate a PDF in the style of worksheet which includes question sets covering each section of the chapter in the given file.",
                },
                {
                    "type": "document",
                    "source": {"type": "file", "file_id": uploaded_id},
                },
            ],
        }
    ],
    container={
        "skills":[
            {
                "type":"anthropic",
                "skill_id":"pdf",
                "version":"latest"
            }
        ]
    },
     tools=[
        {
            "type": "code_execution_20260521",
            "name": "code_execution"
        }
    ]
)

pprint.pprint(response.model_dump_json(indent=2))


out_dir = Path("generated")
out_dir.mkdir(exist_ok=True)

for block in response.content:
    if block.type != "bash_code_execution_tool_result":
        continue
    result = block.content
    if result.type != "bash_code_execution_result":
        continue
    for output in result.content:
        meta = client.files.retrieve_metadata(output.file_id)
        path = out_dir / (meta.filename or f"{output.file_id}.pdf")
        client.files.download(output.file_id).write_to_file(path)
        print(path, meta.mime_type, meta.size_bytes)
```

```python
messages=[]

def add_user_message(messages, message):
    messages.append({
        "role":"user",
        "content": message
    })
    
def add_assistant_message(messages, message):
    messages.append({
        "role":"assistant",
        "content": message
    })
    
    
def chat(messages):
    answer = client.messages.create(model = model, max_tokens=1024, messages=messages, thinking={"type":"between_tools"})
    return answer.content[0].text


# add_user_message(messages, "What is 2*2?")
# messages
# answer = chat(messages)
# add_assistant_message(messages, answer)
# messages


def chat_bot():
    while True:
        user_input = input("You: ")
        if user_input.lower() in ["exit", "quit"]:
            print("Bot: Goodbye!")
            break
        add_user_message(messages, user_input)
        answer = chat(messages)
        add_assistant_message(messages, answer)
        print("You: ", user_input)
        print(f"Bot: {answer}")


chat_bot()
```

```python
from anthropic import beta_tool
import json
import random

def _weather(location: str) -> dict:
    descriptions ={
        0: "snowy" 
    }
    for i in range(1, 11):
        descriptions[i]= "cold"
    for i in range(11, 16):
        descriptions[i]= "cloudy"
    for i in range(16, 21):
        descriptions[i]= "rainy"
    for i in range(21, 31):
        descriptions[i]= "sunny"
    random_temp = random.randint(0, 30)
    description = descriptions.get(random_temp)
    return {
        "location": location,
        "temperature": f"{random_temp}C",
        "description": description
    }

@beta_tool
def get_weather(location: str) -> str:
    """Get the weather for a given location"""
    return json.dumps(_weather(location))

@beta_tool
def predict_weather(location: str, days: int) -> str:
    """Predict the weather for a given location and number of days"""
    return json.dumps([_weather(location) for _ in range(days)])

runner =client.beta.messages.tool_runner(
    model = model, 
    max_tokens = 1024,
    tools= [get_weather, predict_weather],
    messages= [
        {
            "role": "user",
            "content": "What is the weather in Tokyo for the next week?"
        }
    ]
)

for message in runner:
    if getattr(message,'content', None):
        content = message.content
        for block in content:
            if block.type == 'text':
                print(block.text, end='')
            elif block.type == 'tool_use':
                print(f"\n{block.name} called with {block.input}")
            elif block.type =='thinking':
                print(block.thinking, end='')
                    
    
    
    
    
    
```
